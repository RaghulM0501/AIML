# Methods

```python
class BankAccount:
    bank_name = "Global Bank"
    
    def __init__(self, account_holder, balance = 0):
        self.account_holder = account_holder  #instance attribute
        self.balance = balance                #instance attribute
    
    def deposit(self, amount):                #instance method  
        self.balance += amount
        bank_name = "RRR"
        return f"Deposited {amount}. New balance is {self.balance}."
    
    def withdraw(self, amount):               #instance method
        if amount > self.balance:
            return "Insufficient funds."
        self.balance -= amount
        return f"Withdrew {amount}. New balance is {self.balance}."
    
    @classmethod
    def change_bank_name(cls, new_name):  #class method
        cls.bank_name = new_name
    
    @staticmethod
    def bank_policy():                     #static method
        return "All accounts must maintain a minimum balance of $100."
        
account1 = BankAccount("Alice", 1000)         #instance method; account1 is an object of BankAccount class  
account2 = BankAccount("Bob", 500)
account1.deposit(200)
print(account1.balance)
```

# Inheritance

```python
class Animal:
    def eat(self):
        return "The animal is eating."
class Dog(Animal):
    def bark(self):
        return "Woof! Woof!"
```

```python
d = Dog()
```

```python
dir(d)
```

```python
d.bark()
```

```python
d.eat()
```

```python
isinstance(d, Dog)
```

```python
isinstance(d, Animal)
```

```python
issubclass(Dog, Animal)
```

```python
issubclass(Animal, Dog)
```

# Multiple inheritance

```python
class A:
    def method_a(self):
        return "A"
class B:
   def method_b(self):
       return "B"
class C(A, B):
    def method_c(self):
        pass
```

```python
c = C()
```

```python
c.method_a()
c.method_b()
```

```python
c.method_a()
```

# Multilevel Inheritance

```python
class A:
    def method_a(self):
        return "A"
class B(A):
   def method_b(self):
       return "B"
class C(B):
    def method_c(self):
        pass
```

```python
c = C()
c.method_a()
```

# Herarchical Inheritance

```python
class Animal:
    def eat(self):
        return "The animal is eating."
class Dog(Animal):
    def bark(self):
        return "Woof! Woof!"
class Cat(Animal):
    def meow(self):
        return "Meow! Meow!"
```

```python
c = Cat()
c.meow()
```

```python
c.eat()
```

```python
c.bark()
```

# Hybrid Inheritance

```python
class A:
    pass
class B(A):
    pass
class C(A):
    pass
class D(B, C):
    pass
```

```python
d = D() # the order of execution will be D->B->C->A
```

```python
print(D.mro()) # This method shows the method resolution order
```
