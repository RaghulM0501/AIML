# Series:
- It is a 1D data 
- Like list

```python
data_1D = [10, 20, 30, 40, 50]
series = pd.Series(data_1D)
```

```python
series
```

Labelling a dataset

```python
data_1D = [10, 20, 30, 40, 50]
series = pd.Series(data_1D, index=['a', 'b', 'c', 'd', 'e'])
```

```python
series
```

Dictionary doesn't need indexing as the key will be automatically considered as INDEX

```python
data_1D = { 'Name': 'Alice', 'Age': 30, 'City': 'New York'}
series = pd.Series(data_1D)
```

```python
series
```

```python
series["Name"]
```
